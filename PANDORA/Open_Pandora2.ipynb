```python
def pandora_quicklook(pandora_id,sitename,data_code,start,end):
'''
INPUTS:
pandora_id: ID Number of the Pandora Instrument {int}
sitename: Location Name of the Pandora Instrument {string}
start: Date and Time of the First Desired Observations {string}, yyyy-mm-ddTHH:MM:SS
end: Date and Time of the Last Desired Observations {string}, yyyy-mm-ddTHH:MM:SS
## NOTE THE REQUIRED 'T' BETWEEN THE DATE AND HOUR ##
data_code: 5 Character String Identifying the Trace Gas and Profile {string}
    data_code is one of the following
        rfuh5: Formaldehyde vertical profiles
        rfus5: Formaldehyde total column
        rnvh3: Nitrogen Dioxide vertical profiles
        rnvs3: Nitrogen Dioxide total column
        rout2: Ozone total column
        rsus1: Sulfur Dioxide total column
        rwvt1: Water Vapor total column
OUTPUTS:
Pandas Dataframe with the requested data
'''
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import requests
#from Pandora API Tutorial: https://api.pandonia-global-network.org/v1/tutorial/
BASE_URL = "https://api.pandonia-global-network.org/v1" 
bulk_url = f"{BASE_URL}/download/bulk_l2"
params = { #set parameters for bulk download, using values input into this function
    "pan_id": pandora_id,
    "location": sitename,
    "code": data_code,
    "start_datetime": start,
    "end_datetime": end
}
response = requests.get(bulk_url, params=params)

if response.status_code == 200:
    filename = response.headers.get('Content-Disposition').split('filename=')[1]
    print(filename)
    with open(filename, "wb") as f:
        f.write(response.content)
    print(f"Downloaded bulk L2 file: {filename}")
else:
    print(f"Error: {response.status_code}")
#END API Tutorial
 
for id in ['rfuh5','rfus5','rnvh3','rnvs3','rout2','rsus1','rwvt1']:
    if id in filename:
        fileid = id

#open the .txt file using the found parameters
pandora_df=pd.read_csv(filename, sep = " ",encoding='latin',on_bad_lines='warn', 
               skiprows=hard_set_values[fileid][0], names=columns[fileid])

#catch here - if there is no data, end the function
if len(pandora_df) == 0:
    print("Data File is Empty")
    return 
 
#some quality control aspects - remove all values below 0 (unphysical) and any values flagged by DQ2
## IF DQ2 > 20 - data is unusable
pandora_df = pandora_df[pandora_df.iloc[:,hard_set_values[fileid][2]].values<20]

#catch here - if there is no data, end the function
if len(pandora_df) == 0:
    print("Data File is Empty")
    return 

#return the dataframe with requested data
return pandora_df
```
